**Q1. Pearson correlation coefficient is a measure of the linear relationship between two variables. Suppose
you have collected data on the amount of time students spend studying for an exam and their final exam
scores. Calculate the Pearson correlation coefficient between these two variables and interpret the result.**

The Pearson correlation coefficient measures the strength and direction of the linear relationship between two continuous variables. The formula for the Pearson correlation coefficient $ r $ is:

$ r = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum (x_i - \bar{x})^2 \sum (y_i - \bar{y})^2}} $

where:
- $ x_i $ and $ y_i $ are the individual sample points for variables $ x $ and $ y $.
- $ \bar{x} $ and $ \bar{y} $ are the means of the variables $ x $ and $ y $.

Let's calculate the Pearson correlation coefficient between the amount of time students spend studying for an exam and their final exam scores using a sample dataset.

### Step-by-Step Calculation

1. **Create a Sample Dataset**

Let's create a sample dataset with two variables: `study_time` (hours) and `exam_score` (percentage).

```python
import pandas as pd
import numpy as np

# Sample data
data = {
    'study_time': [2, 3, 5, 7, 9, 4, 6, 8, 3, 5],
    'exam_score': [50, 55, 65, 70, 85, 60, 75, 80, 55, 68]
}

df = pd.DataFrame(data)
```

2. **Calculate the Pearson Correlation Coefficient**

We can use the `pearsonr` function from the `scipy.stats` module to calculate the Pearson correlation coefficient.

```python
from scipy.stats import pearsonr

# Calculate the Pearson correlation coefficient
corr, _ = pearsonr(df['study_time'], df['exam_score'])
print(f'Pearson correlation coefficient: {corr:.3f}')
```

### Interpretation

Let's assume the output is:

```
Pearson correlation coefficient: 0.923
```

### Interpreting the Result

1. **Magnitude**: The Pearson correlation coefficient ranges from -1 to 1.
   - A value of 1 indicates a perfect positive linear relationship.
   - A value of -1 indicates a perfect negative linear relationship.
   - A value of 0 indicates no linear relationship.

2. **Direction**: 
   - A positive value (0.923) indicates a positive linear relationship between study time and exam scores. As the amount of time spent studying increases, the exam scores tend to increase as well.
   - A negative value would indicate that as one variable increases, the other decreases.

3. **Strength**:
   - A coefficient of 0.923 suggests a very strong positive linear relationship between the variables. This means that the amount of time students spend studying is strongly associated with higher exam scores.

### Full Code Example

Here is the complete code for calculating the Pearson correlation coefficient and interpreting the result:

```python
import pandas as pd
import numpy as np
from scipy.stats import pearsonr

# Sample data
data = {
    'study_time': [2, 3, 5, 7, 9, 4, 6, 8, 3, 5],
    'exam_score': [50, 55, 65, 70, 85, 60, 75, 80, 55, 68]
}

df = pd.DataFrame(data)

# Calculate the Pearson correlation coefficient
corr, _ = pearsonr(df['study_time'], df['exam_score'])
print(f'Pearson correlation coefficient: {corr:.3f}')
```

This calculation demonstrates that there is a very strong positive linear relationship between the amount of time students spend studying and their final exam scores, indicating that increased study time is associated with higher exam performance.

**Q2. Spearman's rank correlation is a measure of the monotonic relationship between two variables.
Suppose you have collected data on the amount of sleep individuals get each night and their overall job
satisfaction level on a scale of 1 to 10. Calculate the Spearman's rank correlation between these two
variables and interpret the result.**

Spearman's rank correlation coefficient is a non-parametric measure of the strength and direction of the monotonic relationship between two variables. Unlike the Pearson correlation coefficient, it does not assume that the relationship between the variables is linear. Instead, it assesses how well the relationship between two variables can be described using a monotonic function.

The formula for Spearman's rank correlation coefficient $ \rho $ is:

$ \rho = 1 - \frac{6 \sum d_i^2}{n(n^2 - 1)} $

where:
- $ d_i $ is the difference between the ranks of the $i$-th observation for the two variables.
- $ n $ is the number of observations.

### Step-by-Step Calculation

1. **Create a Sample Dataset**

Let's create a sample dataset with two variables: `sleep_hours` (hours per night) and `job_satisfaction` (satisfaction level on a scale of 1 to 10).

```python
import pandas as pd
import numpy as np

# Sample data
data = {
    'sleep_hours': [7, 8, 6, 5, 7, 8, 9, 6, 5, 7],
    'job_satisfaction': [6, 7, 5, 4, 6, 8, 9, 5, 4, 6]
}

df = pd.DataFrame(data)
```

2. **Calculate the Spearman's Rank Correlation Coefficient**

We can use the `spearmanr` function from the `scipy.stats` module to calculate Spearman's rank correlation coefficient.

```python
from scipy.stats import spearmanr

# Calculate the Spearman's rank correlation coefficient
corr, _ = spearmanr(df['sleep_hours'], df['job_satisfaction'])
print(f'Spearman\'s rank correlation coefficient: {corr:.3f}')
```

### Interpretation

Let's assume the output is:

```
Spearman's rank correlation coefficient: 0.915
```

### Interpreting the Result

1. **Magnitude**: The Spearman's rank correlation coefficient ranges from -1 to 1.
   - A value of 1 indicates a perfect positive monotonic relationship.
   - A value of -1 indicates a perfect negative monotonic relationship.
   - A value of 0 indicates no monotonic relationship.

2. **Direction**:
   - A positive value (0.915) indicates a positive monotonic relationship between sleep hours and job satisfaction. As the amount of sleep increases, job satisfaction tends to increase as well.
   - A negative value would indicate that as one variable increases, the other decreases.

3. **Strength**:
   - A coefficient of 0.915 suggests a very strong positive monotonic relationship between the variables. This means that the amount of sleep individuals get each night is strongly associated with higher job satisfaction levels.

### Full Code Example

Here is the complete code for calculating the Spearman's rank correlation coefficient and interpreting the result:

```python
import pandas as pd
import numpy as np
from scipy.stats import spearmanr

# Sample data
data = {
    'sleep_hours': [7, 8, 6, 5, 7, 8, 9, 6, 5, 7],
    'job_satisfaction': [6, 7, 5, 4, 6, 8, 9, 5, 4, 6]
}

df = pd.DataFrame(data)

# Calculate the Spearman's rank correlation coefficient
corr, _ = spearmanr(df['sleep_hours'], df['job_satisfaction'])
print(f'Spearman\'s rank correlation coefficient: {corr:.3f}')
```

This calculation demonstrates that there is a very strong positive monotonic relationship between the amount of sleep individuals get each night and their overall job satisfaction level, indicating that more sleep is associated with higher job satisfaction.

**Q3. Suppose you are conducting a study to examine the relationship between the number of hours of
exercise per week and body mass index (BMI) in a sample of adults. You collected data on both variables
for 50 participants. Calculate the Pearson correlation coefficient and the Spearman's rank correlation
between these two variables and compare the results.**

To calculate the Pearson correlation coefficient and Spearman's rank correlation coefficient between the number of hours of exercise per week and body mass index (BMI), follow these steps:

### Pearson Correlation Coefficient (r)

The Pearson correlation coefficient measures the linear relationship between two continuous variables.

1. **Calculate the means and standard deviations:**
   - Let $ X $ denote the number of hours of exercise per week.
   - Let $ Y $ denote the BMI.

   Calculate the means:
   $
   \bar{X} = \frac{\sum X}{n}, \quad \bar{Y} = \frac{\sum Y}{n}
   $

   Calculate the standard deviations:
   $
   s_X = \sqrt{\frac{\sum (X - \bar{X})^2}{n-1}}, \quad s_Y = \sqrt{\frac{\sum (Y - \bar{Y})^2}{n-1}}
   $

2. **Calculate the Pearson correlation coefficient $ r $:**
   $
   r = \frac{\sum (X_i - \bar{X})(Y_i - \bar{Y})}{\sqrt{\sum (X_i - \bar{X})^2 \sum (Y_i - \bar{Y})^2}}
   $

   Where $ X_i $ and $ Y_i $ are individual data points.

### Spearman's Rank Correlation Coefficient $( \rho )$

Spearman's rank correlation coefficient assesses the monotonic relationship between two variables, based on their ranks rather than their raw values.

1. **Rank the data:**
   - Rank the values of $ X $ and $ Y $ separately. If ties occur, assign each tied value the average of the ranks they would have received had they not been tied.

2. **Calculate Spearman's $ \rho $:**
   $
   \rho = 1 - \frac{6 \sum d_i^2}{n(n^2 - 1)}
   $
   
   Where $ d_i $ is the difference between the ranks of corresponding pairs of $ X $ and $ Y $.

### Comparison of Results

- **Interpretation:**
  - Pearson correlation $ r $ measures the strength and direction of the linear relationship between exercise hours and BMI.
  - Spearman's $ \rho $ assesses the monotonic relationship (whether increasing or decreasing) between ranks of exercise hours and BMI.
  
- **When to Use:**
  - Use Pearson correlation when the relationship between variables is expected to be linear.
  - Use Spearman's correlation when the relationship may be monotonic but not necessarily linear, or when the data are ranked rather than on a numerical scale.

- **Strengths and Considerations:**
  - Pearson correlation is sensitive to outliers and assumes normality.
  - Spearman's correlation is more robust to outliers and does not assume normality, but it may lose information about the magnitude of differences between data points.
